```python
text_samples = ["Yesterday I bought a coffee", "Today I went to buy a new phone", "I spent 12 euros for transport ticket", "I spent 1000 euros at clinic today"]
```

```python
# convert to lower text
lower_text = [text.lower() for text in text_samples]
lower_text
```

```python
# tokenization
from nltk.tokenize import sent_tokenize, word_tokenize
import nltk
nltk.download('punkt_tab')
```

```python
sent_tok = [sent_tokenize(sent) for sent in lower_text]
sent_tok
```

```python
word_tok = [word_tokenize(word) for word in lower_text]
word_tok
```

```python
import re

clean_text_3 = []

for words in word_tok:
    clean = []
    for w in words:
        res = re.sub(r'[^\w\s]', "", w)
    if res != "":
        clean.append(res)
    clean_text_3.append(clean)

clean_text_3
```
